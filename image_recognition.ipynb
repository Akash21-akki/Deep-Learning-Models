```python
# With Scaled
tf.keras.datasets.mnist.load_data(
    path='mnist.npz'
)
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
assert x_train.shape == (60000, 28, 28)
assert x_test.shape == (10000, 28, 28)
assert y_train.shape == (60000,)
assert y_test.shape == (10000,)
x_train.flatten
x_test.flatten
input_shape=(28,28)
# x_train_flat = x_train.reshape(len(x_train),28*28)
# x_test_flat = x_test.reshape(len(x_test),28*28)
x_train = x_train / 255;
x_test = x_test / 255;
flatten = keras.layers.Flatten();

model = keras.Sequential([
    keras.layers.Input(shape=input_shape),
    flatten,
    keras.layers.Dense(10,activation="sigmoid"),
    keras.layers.Dense(10, activation='softmax')
    ])

model.compile(optimizer="adam",loss='sparse_categorical_crossentropy',metrics=['Accuracy'])
model.fit(x_train,y_train,epochs=5,verbose=1)
model.evaluate(x_test,y_test,verbose=False)
```

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np

# Load data
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

# Model Architecture - Softmax for better classification
model = keras.Sequential([
    keras.layers.Input(shape=(28, 28)),
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation='relu'),
    # keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Train
model.fit(x_train, y_train, epochs=5)

model.predict(x_test[1:2])

# Save as .keras (Native format)
# model.save("mnist_model.keras")
# print("Model Saved Successfully as mnist_model.keras")
```

```python
from fastapi import FastAPI, UploadFile, File
import tensorflow as tf
import numpy as np
from PIL import Image, ImageOps
import io

app = FastAPI()

# Model Load
model = tf.keras.models.load_model("mnist_model.keras")

def preprocess_image(data):
    # Image open and convert to grayscale
    img = Image.open(io.BytesIO(data)).convert('L')

    # MNIST logic: model expects black background and white digit
    img = ImageOps.invert(img)

    img = img.resize((28, 28))
    img_array = np.array(img) / 255.0
    img_array = img_array.reshape(1, 28, 28)
    return img_array

@app.post("/predict")
async def predict(file: UploadFile = File(...)):
    try:
        data = await file.read()
        img = preprocess_image(data)

        prediction = model.predict(img)
        result = np.argmax(prediction)
        confidence = float(np.max(prediction))

        return {
            "status": "success",
            "prediction": int(result),
            "confidence": f"{confidence:.2%}"
        }
    except Exception as e:
        return {"status": "error", "message": str(e)}
```
