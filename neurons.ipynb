```python
import math

inputs = [0.5, 0.8]
weights = [0.4, 0.7]
bias = 0.2

# STEP 1: y = wx + b
z = sum(w * x for w, x in zip(weights, inputs)) + bias
print(f"Step 1 (z): {z}")

# STEP 2: Sigmoid
prediction = 1 / (1 + math.exp(-z))
print(f"Step 2 (Sigmoid): {prediction:.4f}")

# STEP 3:
if prediction >= 0.5:
    final_class = 1
else:
    final_class = 0

print(f"Step 3 (Result): Class {final_class}")
```

```python
class Neuron:
    def __init__(self, weights, bias):
        self.weights = weights
        self.bias = bias
        self.e = 2.718281828459

    def activate(self, inputs):
        # STEP 1: y = wi*xi + b
        z = 0
        for i in range(len(inputs)):
            z += inputs[i] * self.weights[i]
        z += self.bias

        # STEP 2: Sigmoid
        sigmoid_output = 1 / (1 + (self.e ** -z))

        # STEP 3:
        if sigmoid_output >= 0.5:
            return 1, sigmoid_output
        else:
            return 0, sigmoid_output

my_neuron = Neuron(weights=[0.8, 0.4], bias=-0.5)
result, raw_val = my_neuron.activate([1, 0])

print(f"Sigmoid Value: {raw_val}")
print(f"Final Class: {result}")
```
